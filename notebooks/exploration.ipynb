```python
import pandas as pd
import numpy as np
import joblib

model_path = "data/models/logreg_model.pkl"
model = joblib.load(model_path)

coef = model.coef_[0]
features = model.feature_names_in_

df_coef = pd.DataFrame({
    "feature": features,
    "coef": coef,
    "abs_coef": np.abs(coef)
}).sort_values("abs_coef", ascending=False)

print(df_coef)
```
